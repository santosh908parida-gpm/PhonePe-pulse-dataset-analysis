```python
import os
import json
import pandas as pd

base_path = "pulse/data/aggregated/transaction/country/india/state"

rows = []

for state in os.listdir(base_path):

    state_path = os.path.join(base_path, state)

    for year in os.listdir(state_path):

        year_path = os.path.join(state_path, year)

        for file in os.listdir(year_path):

            if file.endswith(".json"):

                quarter = file.replace(".json", "")
                file_path = os.path.join(year_path, file)

                with open(file_path, "r") as f:
                    data = json.load(f)

                for item in data["data"]["transactionData"]:

                    rows.append({
                        "state": state,
                        "year": int(year),
                        "quarter": int(quarter),
                        "category": item["name"],
                        "count": item["paymentInstruments"][0]["count"],
                        "amount": item["paymentInstruments"][0]["amount"]
                    })

transactions_df = pd.DataFrame(rows)
```

```python
base_path = "pulse/data/aggregated/user/country/india/state"

rows = []

for state in os.listdir(base_path):
    state_path = os.path.join(base_path, state)

    for year in os.listdir(state_path):
        year_path = os.path.join(state_path, year)

        for file in os.listdir(year_path):
            if file.endswith(".json"):

                quarter = file.replace(".json", "")
                file_path = os.path.join(year_path, file)

                with open(file_path, "r") as f:
                    data = json.load(f)

                rows.append({
                    "state": state,
                    "year": int(year),
                    "quarter": int(quarter),
                    "registered_users": data["data"]["aggregated"]["registeredCount"]
                })

users_df = pd.DataFrame(rows)
```

```python
base_path = "pulse/data/aggregated/merchant/country/india/state"

rows = []

for state in os.listdir(base_path):
    state_path = os.path.join(base_path, state)

    for year in os.listdir(state_path):
        year_path = os.path.join(state_path, year)

        for file in os.listdir(year_path):
            if file.endswith(".json"):

                quarter = file.replace(".json", "")
                file_path = os.path.join(year_path, file)

                with open(file_path, "r") as f:
                    data = json.load(f)

                rows.append({
                    "state": state,
                    "year": int(year),
                    "quarter": int(quarter),
                    "registered_merchants": data["data"]["aggregated"]["registeredCount"]
                })

merchants_df = pd.DataFrame(rows)
```

```python
merchants_df.head()
```

```python
merchants_df.shape
```

```python
transactions_wide = transactions_df.pivot_table(
    index=["state", "year", "quarter"],
    columns="category",
    values=["count", "amount"]
).reset_index()

transactions_wide.head()
```

```python
transactions_wide.columns = [
    "state", "year", "quarter",
    "p2p_amount", "retail_amount", "utility_amount",
    "p2p_count", "retail_count", "utility_count"
]

transactions_wide.head()
```

```python
master_df = transactions_wide.merge(
    users_df,
    on=["state", "year", "quarter"],
    how="left"
)

master_df = master_df.merge(
    merchants_df,
    on=["state", "year", "quarter"],
    how="left"
)

master_df.head()
```
